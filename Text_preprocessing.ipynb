Remove HTML Tags

```python
import re
def remove_html_tags(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',text)
text=df['review'][1]
text

new_text=remove_html_tags(text)
new_text
```

```python
df['review']=df['review'].apply(remove_html_tags)
```

```python
def remove_url(text):
    pattern=re.compile(r'https?://\$+|www\.\$+')
    return pattern
```

Remove Punctuation

```python
import string,time
exclude=string.punctuation
print(exclude)
def remove_punc(text):
    for char in exclude:
        text=text.replace(char,'')
    return text

start=time.time()
str11='This is a string,with pauncafkd!!!! )))   oajsdso ((( ksfk ))) *** asdjska&&& afkj $$'
new_string=remove_punc(str11)
time1=time.time()-start
print(time1*50000)
new_string
```
