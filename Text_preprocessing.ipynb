# Removing Abbrevations

```python
import urllib.request

# The URL to the "Raw" version of the slang.txt file on GitHub
url = "https://raw.githubusercontent.com/rishabhverma17/sms_slang_translator/master/slang.txt"

# Fetch the data from the URL
response = urllib.request.urlopen(url)
text_data = response.read().decode('utf-8')

# Create an empty dictionary
slang_dict = {}

# Process the text line by line
for line in text_data.strip().split('\n'):
    # Split each line at the first '=' sign
    if '=' in line:
        key, value = line.split('=', 1)
        # Remove any extra whitespace and add it to the dictionary
        slang_dict[key.strip()] = value.strip()

# Print the first 5 items to verify it worked
print(list(slang_dict.items())[:5])
```

```python
def remove_abbrevations(text):
    new_text=[]
    for w in text.split():
        if w.upper() in slang_dict:
            new_text.append(slang_dict[w.upper()])
        else:
            new_text.append(w)
    return ' '.join(new_text)

remove_abbrevations('idk what is going on')
```

# Removing Stopwords

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    return ' '.join(new_text)
output=remove_stopwords('this is a big sentence isnt it?')
print(output)
```

# Removing emojis

```python
def remove_emoji(text):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
        u"\U00002702-\U000027B0"
        u"\U000024C2-\U0001F251"
        "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

sentence='Clg ka naam aa gaya 😂😂'
remove_emoji(sentence)
```
